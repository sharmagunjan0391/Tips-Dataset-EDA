```python
sns.set_theme(style="whitegrid")

plt.figure(figsize=(8, 6))
sns.histplot(df['total_bill'], bins= 20, color = 'skyblue', edgecolor = 'white')

plt.title("Distributions of Total Bills ", fontsize = 14, fontweight= 'bold')
plt.xlabel("Total Bill ($)", fontsize = 12)
plt.ylabel("Frequency", fontsize = 12)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.kdeplot(df['total_bill'], color='darkblue', fill=True, linewidth=2)

plt.title('Density Distribution of Total Bills (KDE)', fontsize=14, fontweight='bold')
plt.xlabel('Total Bill ($)', fontsize=12)
plt.ylabel('Density', fontsize=12)
plt.show()
```

```python
sns.set_theme(style="whitegrid")

# Create a figure with 3 subplots
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 1. Histogram
sns.histplot(df['tip'], bins=15, color='coral', edgecolor='black', ax=axes[0])
axes[0].set_title('Histogram of Tips', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Tip Amount ($)')
axes[0].set_ylabel('Frequency')

# 2. KDE Plot
sns.kdeplot(df['tip'], color='darkred', fill=True, linewidth=2, ax=axes[1])
axes[1].set_title('KDE Plot of Tips', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Tip Amount ($)')
axes[1].set_ylabel('Density')

# 3. Combo Plot (Histogram + KDE)
sns.histplot(df['tip'], kde=True, bins=15, color='crimson', edgecolor='black', ax=axes[2])
axes[2].set_title('Histogram + KDE Overlay', fontsize=12, fontweight='bold')
axes[2].set_xlabel('Tip Amount ($)')
axes[2].set_ylabel('Count')

plt.tight_layout()
plt.show()
```

```python
# Select numerical columns
numeric_df = df.select_dtypes(include="number")

# Calculate correlation
correlation = numeric_df.corr()

# Plot heatmap
plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    square=True
)

plt.title("Correlation Between Numerical Variables", fontsize=16)
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.show()
```

```python
# Create JointGrid
g = sns.JointGrid(
    data=df,
    x="total_bill",
    y="tip",
    hue="time",
    palette="Set2"
)

# Scatterplot
g.plot_joint(
    sns.scatterplot,
    alpha=0.7,
    s=70
)

# Add regression lines separately
for time, color in zip(["Lunch", "Dinner"], sns.color_palette("Set2", 2)):
    data = df[df["time"] == time]
    
    sns.regplot(
        data=data,
        x="total_bill",
        y="tip",
        scatter=False,
        color=color,
        ax=g.ax_joint
    )

# Marginal distributions
g.plot_marginals(
    sns.histplot,
    kde=True,
    alpha=0.4
)

# Title
g.fig.suptitle(
    "Total Bill vs Tip: Lunch vs Dinner",
    y=1.02,
    fontsize=15,
    fontweight="bold"
)

plt.show()
```
